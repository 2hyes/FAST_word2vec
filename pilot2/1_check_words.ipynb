```python
import os
import re

import pandas as pd
import numpy as np
from gensim.models import KeyedVectors
```

```python
pilot2_dir = os.getcwd()

# loading the word2vec model
word2vec_model = KeyedVectors.load_word2vec_format(pilot2_dir + '/../pretrained/GoogleNews-vectors-negative300.bin', binary=True)
# print(word2vec_model['Eiffel']) # 모델이 잘 로드되었는지 확인
```

```python
processed_data_dir = os.getcwd() + '/data/processed/' # data위치 지정

pilot_data = pd.read_csv(processed_data_dir + 'merged_data.csv')
pilot_data[0:3]
```

```python
seed_words = ['key', 'money', 'friend']
n_respond_words = 30 # 하나의 시드당 30개의 단어 응답
n_subject = len(pilot_data) # 58
n_dim_of_vector = 300

except_words = []
for seed_word in seed_words: ##### 차례로 seed_word에 key, money, friend가 저장됨
    word_columns = [seed_word + str(i) for i in range(1, n_respond_words+1)] # key1~30, money1~30, friend1~30

    for column in word_columns: ##### 차례로 column에 데이터의 컬럼 하나씩 저장됨. ex) key1
        for i_subject in range(n_subject):  ##### 차례로 피험자 숫자ID(0~57)
            response_word = pilot_data.iloc[i_subject][column]
            # remove white space after the word such as 'stoma ' and ' stoma'
            pilot_data[column] = pilot_data[column].str.strip()
            # Remove special character
            pilot_data[column] = pilot_data[column].apply(lambda x: re.sub('[^\w\s]', '', str(x)) if isinstance(x, (str, np.ndarray)) else x)
            # Lower case로 통일
            pilot_data[column] = pilot_data[column].str.lower()
        
            print(response_word)
            ## exceptional cases (1차로 뽑아낸 후에, 추가한 코드 부분)
            # typo fix
            # wamrth -> warmth
            pilot_data[column] = pilot_data[column].replace('wamrth', 'warmth')
            # doir -> door
            pilot_data[column] = pilot_data[column].replace('doir', 'door')
            # nneck -> neck
            pilot_data[column] = pilot_data[column].replace('nneck', 'neck')
            # locok -> lock
            pilot_data[column] = pilot_data[column].replace('locok', 'lock')
            # sald -> said
            pilot_data[column] = pilot_data[column].replace('sald', 'said')
            # architecht -> architect
            pilot_data[column] = pilot_data[column].replace('architecht', 'architect')
            # tattoes -> tattoo
            pilot_data[column] = pilot_data[column].replace('tattoes', 'tattoo')
            # horrer -> horror
            pilot_data[column] = pilot_data[column].replace('horrer', 'horror')
            # sissors -> scissors
            pilot_data[column] = pilot_data[column].replace('sissors', 'scissors')
            # estethician -> esthetician
            pilot_data[column] = pilot_data[column].replace('estethician', 'esthetician')
            # breakfeasr -> breakfast
            pilot_data[column] = pilot_data[column].replace('breakfeasr', 'breakfast')
            # quickl -> quickly
            pilot_data[column] = pilot_data[column].replace('quickl', 'quickly')
            # tohether -> together
            pilot_data[column] = pilot_data[column].replace('tohether', 'together')
            # watermelom -> watermelon
            pilot_data[column] = pilot_data[column].replace('watermelom', 'watermelon')
            # dollard -> dollar
            pilot_data[column] = pilot_data[column].replace('dollard', 'dollar')
            # togethere -> together
            pilot_data[column] = pilot_data[column].replace('togethere', 'together')
            # favourite -> favorite
            pilot_data[column] = pilot_data[column].replace('favourite', 'favorite')
            # establisjed -> established
            pilot_data[column] = pilot_data[column].replace('establisjed', 'established')
            # groeries -> groceries
            pilot_data[column] = pilot_data[column].replace('groeries', 'groceries')
            # relationshi -> relationship
            pilot_data[column] = pilot_data[column].replace('relationshi', 'relationship')
            # reaad -> read
            pilot_data[column] = pilot_data[column].replace('reaad', 'read')
            # peoplr -> people
            pilot_data[column] = pilot_data[column].replace('peoplr', 'people')
            # perspm -> person
            pilot_data[column] = pilot_data[column].replace('perspm', 'person')
            # Protecter -> protector
            pilot_data[column] = pilot_data[column].replace('Protecter', 'protector')
            # grey -> gray
            pilot_data[column] = pilot_data[column].replace('grey', 'gray')
            # colour -> color
            pilot_data[column] = pilot_data[column].replace('colour', 'color')
            # comepression -> compression
            pilot_data[column] = pilot_data[column].replace('comepression', 'compression')
            # defenitly -> definitely
            pilot_data[column] = pilot_data[column].replace('defenitly', 'definitely')
            # refress -> refresh
            pilot_data[column] = pilot_data[column].replace('refress', 'refresh')
            # enlighted -> enlightened
            pilot_data[column] = pilot_data[column].replace('enlighted', 'enlightened')
            # claustophobia -> claustrophobia
            pilot_data[column] = pilot_data[column].replace('claustophobia', 'claustrophobia')
            # knwedge -> knowledge
            pilot_data[column] = pilot_data[column].replace('knwedge', 'knowledge')
            # biscute -> biscuit
            pilot_data[column] = pilot_data[column].replace('biscute', 'biscuit')
            # anethestize -> anesthetize
            pilot_data[column] = pilot_data[column].replace('anethestize', 'anesthetize')

            ## capital
            # zurich -> Zurich
            pilot_data[column] = pilot_data[column].replace('zurich', 'Zurich')
            # madagascar -> Madagascar
            pilot_data[column] = pilot_data[column].replace('madagascar', 'Madagascar')
            # eiffel tower -> Eiffel tower
            pilot_data[column] = pilot_data[column].replace('eiffel tower', 'Eiffel tower')
            # sagrada familia -> Sagrada Familia
            pilot_data[column] = pilot_data[column].replace('sagrada familia', 'Sagrada Familia')
            # notredame cathedral -> Notre Dame cathedral
            pilot_data[column] = pilot_data[column].replace('notredame cathedral', 'Notre Dame cathedral')
            # montmartre -> Montmartre
            pilot_data[column] = pilot_data[column].replace('montmartre', 'Montmartre')
            # versailles palace -> Versailles palace
            pilot_data[column] = pilot_data[column].replace('versailles palace', 'Versailles palace')
            # parthenon -> Parthenon
            pilot_data[column] = pilot_data[column].replace('parthenon', 'Parthenon')
            # sydney harbour bridg -> Sydney Harbour Bridge
            pilot_data[column] = pilot_data[column].replace('sydney harbour bridg', 'Sydney Harbour Bridge')
            # sydney harbour bridge -> Sydney Harbour Bridge
            pilot_data[column] = pilot_data[column].replace('sydney harbour bridge', 'Sydney Harbour Bridge')
            # trevi fountain -> Trevi fountain
            pilot_data[column] = pilot_data[column].replace('trevi fountain', 'Trevi fountain')
            # bondi beach -> Bondi beach
            pilot_data[column] = pilot_data[column].replace('bondi beach', 'Bondi beach')
            # machu picchu -> Machu Picchu
            pilot_data[column] = pilot_data[column].replace('machu picchu', 'Machu Picchu')
            # yosemite national pa -> Yosemite National Park
            pilot_data[column] = pilot_data[column].replace('yosemite national pa', 'Yosemite National Park')
            # yosemite national park -> Yosemite National Park
            pilot_data[column] = pilot_data[column].replace('yosemite national park', 'Yosemite National Park')
            # santorini -> Santorini
            pilot_data[column] = pilot_data[column].replace('santorini', 'Santorini')
            # tahiti -> Tahiti
            pilot_data[column] = pilot_data[column].replace('tahiti', 'Tahiti')
            # bitcoin -> Bitcoin
            pilot_data[column] = pilot_data[column].replace('bitcoin', 'Bitcoin')
            # buckingham palace -> Buckingham palace
            pilot_data[column] = pilot_data[column].replace('buckingham palace', 'Buckingham palace')
            # stonehenge -> Stonehenge
            pilot_data[column] = pilot_data[column].replace('stonehenge', 'Stonehenge')
            # instagram -> Instagram
            pilot_data[column] = pilot_data[column].replace('instagram', 'Instagram')
            # tiktok -> Tiktok
            pilot_data[column] = pilot_data[column].replace('tiktok', 'Tiktok')




            ## need to be splited
            # stormcloud -> storm cloud
            pilot_data[column] = pilot_data[column].replace('stormcloud', 'storm cloud')
            # didnt you -> did not
            pilot_data[column] = pilot_data[column].replace('didnt you', 'did not you')
            # aint easy -> be not easy
            pilot_data[column] = pilot_data[column].replace('aint easy', 'be not easy')
            # goodtimes -> good time
            pilot_data[column] = pilot_data[column].replace('goodtimes', 'good time')
            # slowmotion -> slow motion
            pilot_data[column] = pilot_data[column].replace('slowmotion', 'slow motion')
            # lightnight -> light night
            pilot_data[column] = pilot_data[column].replace('lightnight', 'light night')
            
            if isinstance(response_word, str):
                response_word = response_word.split()
                response_word = [response_word for response_word in response_word if response_word not in ['is', 'a','to','of','and']]

                for i_el in range( len(response_word) ):
                    try:
                        vec_word2vec_in = word2vec_model[response_word[i_el]]
                    except:
                        except_word = {
                            'Prolific_ID': pilot_data.iloc[i_subject]['Prolific_ID'],
                            'subject': pilot_data.iloc[i_subject]['subject'],
                            'category': column,
                            'word': response_word
                        }
                        except_words.append(except_word)
```
