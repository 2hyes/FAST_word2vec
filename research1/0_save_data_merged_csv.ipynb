rawdata_study1, rawdata_study2 데이터의 피험자들의 응답결과들을
각 피험자별 하나의 row로 처리하는 전처리 코드입니다.

```python
import os
import glob

import pandas as pd
import numpy as np
```

```python
data_dir = os.getcwd() + '/data/' # data위치 지정
datasets = glob.glob( data_dir + 'raw/' + 'rawdata*.xls' )

n_kind_task = len(datasets)
vec_str_task_name = np.empty( (1,n_kind_task) , dtype='object')
vec_str_task_id = np.empty( (1,n_kind_task)  , dtype='object')
```

```python
total_data_array = []

# rawdata_study1.xls 피험자 137명
for sheet_index in range(1, 138): 
    # 시트 차례로 불러오기
    tbl_data = pd.read_excel(datasets[0], sheet_name=f'Sheet{sheet_index}', header=None)
    # row별로 dict 생성
    tbl_data_dict = tbl_data.to_dict('records') # row별로 dict 생성

    subject_data_dict = {}
    subject_data_dict['subject'] = sheet_index
    for i, answer in enumerate(tbl_data_dict):
        # seedword일 때는 데이터저장하지않고, 컬럼이름 짓기
        if answer[4] == 'seedword':
            key_cnt = 0
            base_col_name = answer[10] # ex. tear_1  
        else:
            key_cnt += 1
            subject_data_dict[base_col_name + str(key_cnt)] = answer[10]
    total_data_array.append(subject_data_dict)
```

```python
# rawdata_study2.xls 피험자 213명
for sheet_index in range(1,214):
    # 시트 차례로 불러오기
    tbl_data = pd.read_excel(datasets[1], sheet_name=f'Sheet{sheet_index}', header=None)
    # row별로 dict 생성
    tbl_data_dict = tbl_data.to_dict('records')
    
    subject_data_dict = {}
    subject_data_dict['subject'] = sheet_index + 137
    for i, answer in enumerate(tbl_data_dict):
        value_type = answer[4].replace("'", "")
        value = answer[10].replace("'", "")

        # seedword일 때는 데이터저장하지않고, 컬럼이름 짓기
        if value_type == 'seedword':
            key_cnt = 0
            base_col_name = value # ex. tear_1
        else:
            if value == '%-NA-%': # null값 처리
                value = ''
            key_cnt += 1
            subject_data_dict[base_col_name + str(key_cnt)] = value
    total_data_array.append(subject_data_dict)
```

```python
pd.DataFrame(total_data_array).to_csv(data_dir + 'processed/merged_data.csv', index=None) 
```
